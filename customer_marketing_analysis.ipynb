```python
purchase_data = {
    'Type': ['Web Purchases', 'Store Purchases'],
    'Average': [df['NumWebPurchases'].mean(),
                df['NumStorePurchases'].mean()]
}

purchase_df = pd.DataFrame(purchase_data)

sns.barplot(x='Type', y='Average', data=purchase_df)

plt.title("Average Web vs Store Purchases")

plt.show()
```

Customers make more purchases in-store than online on average, indicating that physical retail channels remain an important driver of customer activity. Expanding omnichannel strategies and improving the online shopping experience may help increase digital sales while maintaining strong in-store engagement.
